The garment manufacturing industry is one of the largest contributors to the global fashion supply chain. Maintaining high productivity while ensuring quality, efficient workforce utilization, and timely production remains a significant challenge for garment manufacturers.

This project focuses on analyzing worker productivity in a garment manufacturing environment using a real-world dataset. By applying data analytics techniques, the study aims to identify the key factors influencing productivity, evaluate operational performance, and generate meaningful business insights.

The analysis includes data cleaning, exploratory data analysis (EDA), visualization, and interpretation of production-related variables such as overtime, idle time, incentives, work in progress (WIP), standard minute value (SMV), and actual productivity. Garment manufacturing companies often experience fluctuations in worker productivity due to multiple operational factors such as overtime, idle time, production planning, workforce allocation, and work-in-progress (WIP).

This project aims to analyze these factors using real-world production data and provide meaningful insights and recommendations to improve workforce productivity and manufacturing efficiency.

### (v)Feature Engineering ###

Feature engineering is the process of creating new meaningful features from the existing dataset. These newly created categorical attributes simplify complex numerical data, improve interpretability, and support more effective visualization and business analysis.

#### (1) Creating Productivity Level ####

The actual productivity values are categorized into **Low**, **Medium**, and **High** productivity levels. This makes productivity analysis easier and improves the interpretability of visualizations.

```python
df['productivity_level'] = pd.cut(
    df['actual_productivity'],
    bins=[0, 0.5, 0.8, 1.2],
    labels=['Low', 'Medium', 'High']
)

df[['actual_productivity', 'productivity_level']].head()
```

**Observation:-** A new categorical feature named **Productivity_Level** has been created by classifying employees based on their actual productivity performance.

### (2) Creating Overtime Category ###

```python
df['overtime_category'] = pd.cut(
    df['over_time'],
    bins=[0, 5000, 10000, df['over_time'].max()],
    labels=['Regular Hours', 'Extended Hours', 'Excess Hours'],
    include_lowest=True
)
df[['over_time', 'overtime_category']].head()
```

**Observation:-** A new categorical feature named **Overtime_Category** has been created by grouping overtime hours into regular,extended,and excess hours categories. This makes overtime analysis more meaningful and easier to visualize.

**Info:-** Categorizing overtime hours simplifies numerical data into meaningful groups, making it easier to compare productivity across different overtime levels.

### (3) Creating Idle Time Category ###

```python
df['idle_time_category'] = pd.cut(
    df['idle_time'],
    bins=[-0.1, 0, 30, df['idle_time'].max()],
    labels=['No Delay', 'Moderate Delay', 'High Delay']
)

df[['idle_time', 'idle_time_category']].head()
```

**Observation:-** A new feature named **Idle_Time_Category** has been created by grouping idle time into No Delay, Moderate Delay, and High Delay categories. This simplifies the analysis of production delays.

**Info:-** Categorizing idle time makes it easier to analyze the effect of production delays on worker productivity and compare performance across different delay levels.

### (4) Creating Incentive Level


```python
df['incentive_level'] = pd.cut(
    df['incentive'],
    bins=[-0.1, 0, 30, df['incentive'].max()],
    labels=['No Incentive', 'Standard Incentive', 'High Incentive']
)

df[['incentive', 'incentive_level']].head()
```

**Observation:-** A new feature named **Incentive_Level** has been created by classifying employees based on the incentive amount received. This categorization makes it easier to compare productivity across different incentive levels.

**Info:-** Categorizing incentive values helps analyze whether financial rewards contribute to higher employee productivity and operational efficiency.

### (5)Creating SMV Category

```python
df['SMV_category'] = pd.cut(
    df['smv'],
    bins=[0, 10, 20, df['smv'].max()],
    labels=['Simple Task', 'Moderate Task', 'Complex Task'],
    include_lowest=True
)

df[['smv', 'SMV_category']].head()
```

**Observation:-** A new feature named **SMV_Category** has been created by grouping Standard Minute Values into three task complexity levels. This supports productivity analysis based on task complexity.

**Info:-** Categorizing SMV simplifies the analysis of production tasks and helps determine whether more complex tasks have an impact on employee productivity.

### Verify Newly Created Features

```python
df[['productivity_level',
    'overtime_category',
    'idle_time_category',
    'incentive_level',
    'SMV_category']].head()
```

**Observation:-** The newly created categorical features have been successfully added to the dataset. These features will support meaningful visualizations, business insights, and comparative analysis in the subsequent stages of the project.

## 1. Distribution of Actual Productivity using Histogram

- Most productivity values are concentrated between **0.70 and 0.90**, with the highest frequency around **0.80**.
- Very low productivity values (below 0.50) and very high productivity values (above 1.00) occur less frequently.
- The distribution indicates that the majority of workers achieve moderate to high productivity levels, while only a few observations lie at the extreme ends.

### Business Insight

The concentration of actual productivity around **0.80** indicates that most garment workers consistently achieve a satisfactory level of productivity. The small number of low-productivity observations may represent underperforming teams or operational challenges that require further investigation. ## 4. Average Actual Productivity by Day (Bar Plot)

This bar plot compares the **average actual productivity** across different working days. It helps identify whether productivity varies throughout the week and highlights the days with the highest and lowest average productivity.

```python
plt.figure(figsize=(8,5))

day_productivity = (
    df.groupby('day')['actual_productivity']
      .mean()
      .reset_index()
)

# Arrange days in correct order
day_order = [
    'Monday',
    'Tuesday',
    'Wednesday',
    'Thursday',
    'Friday',
    'Saturday',
    'Sunday'
]

sns.barplot(
    data=day_productivity,
    x='day',
    y='actual_productivity',
    order=day_order,
    color='mediumpurple'
)

plt.title('Average Actual Productivity by Day', fontsize=18, fontweight='bold')
plt.xlabel('Day', fontsize=13)
plt.ylabel('Average Actual Productivity', fontsize=13)

plt.grid(axis='y', alpha=0.3)

plt.show()
```

### Observation

- The bar plot compares the average actual productivity across the six working days available in the dataset.
- Average actual productivity remains relatively consistent, ranging from approximately **0.72 to 0.75**.
- **Saturday** records the highest average actual productivity, while **Thursday** has the lowest average productivity among the observed days.
- The variation in productivity across the working days is minimal, indicating stable production performance throughout the week.
- The dataset contains observations for **six working days only**, and no records are available for **Friday**..

### Business Insight

The analysis shows that average productivity remains stable across the recorded working days, with only minor fluctuations. Since productivity does not vary significantly by day, operational factors such as workforce management, overtime, production planning, and resource allocation are likely to have a greater influence on productivity than the day of the week itself.
